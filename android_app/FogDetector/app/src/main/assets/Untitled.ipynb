```python
import tensorflow as tf

interpreter = tf.lite.Interpreter(model_path="fog_model.tflite")
interpreter.allocate_tensors()

print("=== ALL OUTPUT TENSORS ===")
for i, od in enumerate(interpreter.get_output_details()):
    total_bytes = 1
    for dim in od['shape']:
        total_bytes *= dim
    import numpy as np
    dtype_size = np.dtype(od['dtype']).itemsize
    print(f"\nOutput {i}:")
    print(f"  name      : {od['name']}")
    print(f"  shape     : {od['shape']}")
    print(f"  dtype     : {od['dtype']}")
    print(f"  itemsize  : {dtype_size} bytes")
    print(f"  total     : {total_bytes * dtype_size} bytes")
    print(f"  quant     : scale={od['quantization'][0]:.6f}  zero={od['quantization'][1]}")
```
